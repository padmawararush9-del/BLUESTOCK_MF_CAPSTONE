```python
import pandas as pd

# ==========================================================
# LOAD DATA
# ==========================================================

fund_master = pd.read_csv("data/raw/01_fund_master.csv")
nav_history = pd.read_csv("data/raw/02_nav_history.csv")

# ==========================================================
# TASK 6 - FUND MASTER EXPLORATION
# ==========================================================

print("=" * 70)
print("TASK 6 - FUND MASTER EXPLORATION")
print("=" * 70)

print(f"\nTotal Schemes : {len(fund_master)}")

# ---------------------------
# Fund Houses
# ---------------------------

print("\nUnique Fund Houses:")
print("-" * 50)

fund_house_counts = fund_master["fund_house"].value_counts()

for fund, count in fund_house_counts.items():
    print(f"{fund:<40} {count}")

# ---------------------------
# Categories
# ---------------------------

print("\nCategories:")
print("-" * 50)

category_counts = fund_master["category"].value_counts()

for category, count in category_counts.items():
    print(f"{category:<25} {count}")

# ---------------------------
# Sub Categories
# ---------------------------

print("\nSub Categories:")
print("-" * 50)

sub_category_counts = fund_master["sub_category"].value_counts()

for sub_category, count in sub_category_counts.items():
    print(f"{sub_category:<35} {count}")

# ---------------------------
# Risk Categories
# ---------------------------

print("\nRisk Categories:")
print("-" * 50)

risk_counts = fund_master["risk_category"].value_counts()

for risk, count in risk_counts.items():
    print(f"{risk:<25} {count}")

# ---------------------------
# Additional Insights
# ---------------------------

print("\nAdditional Insights:")
print("-" * 50)

print(f"Total Fund Houses       : {fund_master['fund_house'].nunique()}")
print(f"Total Categories        : {fund_master['category'].nunique()}")
print(f"Total Sub Categories    : {fund_master['sub_category'].nunique()}")
print(f"Total Risk Categories   : {fund_master['risk_category'].nunique()}")

# ==========================================================
# TASK 7 - AMFI CODE VALIDATION
# ==========================================================

print("\n")
print("=" * 70)
print("TASK 7 - AMFI CODE VALIDATION")
print("=" * 70)

master_codes = set(fund_master["amfi_code"])
nav_codes = set(nav_history["amfi_code"])

matching_codes = master_codes & nav_codes
missing_in_nav = master_codes - nav_codes
extra_in_nav = nav_codes - master_codes

print(f"\nTotal AMFI Codes in Fund Master : {len(master_codes)}")
print(f"Total AMFI Codes in NAV History : {len(nav_codes)}")

print(f"\nMatching Codes                  : {len(matching_codes)}")
print(f"Missing Codes in NAV History    : {len(missing_in_nav)}")
print(f"Extra Codes in NAV History      : {len(extra_in_nav)}")

coverage = (len(matching_codes) / len(master_codes)) * 100

print(f"\nCoverage Percentage             : {coverage:.2f}%")

# ----------------------------------------------------------
# Missing Codes Details
# ----------------------------------------------------------

if len(missing_in_nav) > 0:

    print("\nMissing AMFI Codes:")
    print("-" * 50)

    missing_funds = fund_master[
        fund_master["amfi_code"].isin(missing_in_nav)
    ][["amfi_code", "scheme_name"]]

    print(missing_funds.to_string(index=False))

else:
    print("\n✓ All AMFI Codes from fund_master are present in nav_history.")

# ----------------------------------------------------------
# Validation Summary
# ----------------------------------------------------------

print("\n")
print("=" * 70)
print("VALIDATION SUMMARY")
print("=" * 70)

if coverage == 100:

    print("✓ Data Validation Successful")
    print("✓ All AMFI codes are mapped correctly")
    print("✓ No missing scheme records detected")
    print("✓ Dataset is ready for analysis, SQL queries and dashboarding")

else:

    print("⚠ Data Validation Failed")
    print("⚠ Some AMFI codes are missing in NAV History")
    print("⚠ Further investigation required")

print("\nNotebook Analysis Complete ✓")
```
