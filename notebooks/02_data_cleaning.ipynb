```python
import pandas as pd
nav = pd.read_csv("../data/raw/02_nav_history.csv")
nav["date"] = pd.to_datetime(nav["date"])
nav = nav.sort_values(
    ["amfi_code", "date"]
)
nav["nav"] = nav.groupby(
    "amfi_code"
)["nav"].ffill()
nav = nav.drop_duplicates()
nav = nav[nav["nav"] > 0]
nav.to_csv(
    "../data/processed/clean_nav.csv",
    index=False
)
```

```python
import pandas as pd
perf = pd.read_csv("../data/raw/07_scheme_performance.csv")
return_cols = [
    "return_1yr_pct",
    "return_3yr_pct",
    "return_5yr_pct",
    "benchmark_3yr_pct"
]

for col in return_cols:
    perf[col] = pd.to_numeric(
        perf[col],
        errors="coerce"
    )
negative_sharpe = perf[
    perf["sharpe_ratio"] < 0
]
expense_outliers = perf[
    (perf["expense_ratio_pct"] < 0.1) |
    (perf["expense_ratio_pct"] > 2.5)
]
perf = perf.drop_duplicates()
perf.to_csv(
    "../data/processed/clean_performance.csv",
    index=False
)
print("\nSaved File:")
print("../data/processed/clean_performance.csv")
```

```python
import pandas as pd
trans = pd.read_csv("../data/raw/08_investor_transactions.csv")
trans["transaction_date"] = pd.to_datetime(
    trans["transaction_date"],
    errors="coerce"
)
trans["transaction_type"] = (
    trans["transaction_type"]
    .str.strip()
    .str.title()
)

trans = trans[
    trans["amount_inr"] > 0
]

trans = trans.drop_duplicates()

valid_kyc = ["Verified", "Pending"]

trans = trans[
    trans["kyc_status"].isin(valid_kyc)
]
trans.to_csv(
    "../data/processed/clean_transactions.csv",
    index=False
)
```
