```python
import pandas as pd

fund_master = pd.read_csv("data/raw/01_fund_master.csv")

fund_master["launch_date"] = pd.to_datetime(
    fund_master["launch_date"],
    errors="coerce"
)

fund_master = fund_master.drop_duplicates()

fund_master.to_csv(
    "data/processed/fund_master_clean.csv",
    index=False
)

print("fund_master_clean.csv created")
```

```python
nav_history = pd.read_csv("data/raw/02_nav_history.csv")

nav_history["date"] = pd.to_datetime(
    nav_history["date"],
    dayfirst=True,
    errors="coerce"
)

nav_history["nav"] = pd.to_numeric(
    nav_history["nav"],
    errors="coerce"
)

nav_history = nav_history.drop_duplicates()

nav_history.to_csv(
    "data/processed/nav_history_clean.csv",
    index=False
)

print("nav_history_clean.csv created")
```

```python
import pandas as pd

fund_master = pd.read_csv("data/raw/01_fund_master.csv")
nav_history = pd.read_csv("data/raw/02_nav_history.csv")

master_codes = set(fund_master["amfi_code"])
nav_codes = set(nav_history["amfi_code"])

matching_codes = master_codes & nav_codes

coverage = (
    len(matching_codes) / len(master_codes)
) * 100
```

```python
validation_df = pd.DataFrame({
    "Total_Fund_Master_Codes":[len(master_codes)],
    "Total_NAV_Codes":[len(nav_codes)],
    "Matching_Codes":[len(matching_codes)],
    "Coverage_Percentage":[coverage]
})

validation_df.to_csv(
    "data/processed/amfi_validation_report.csv",
    index=False
)

print("amfi_validation_report.csv created")
```
